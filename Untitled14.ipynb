```python
import zipfile
import os
import pandas as pd

zip_path = '/content/archive.zip'
extract_path = '/content/amazon_data'


if os.path.exists(zip_path):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)
    print("Extraction complete. Files:", os.listdir(extract_path))
else:
    print("archive.zip not found!")


csv_files = [f for f in os.listdir(extract_path) if f.endswith('.csv')]
if csv_files:
    csv_path = os.path.join(extract_path, csv_files[0])
    df = pd.read_csv(csv_path)
    print(f"Loaded {csv_files[0]} successfully.")
    print(f"Shape: {df.shape}")
    display(df.head())
else:
    print("No CSV file found in extracted contents.")
```

```python
import numpy as np
import pandas as pd

print("Initial Info:")
print(df.info())


df['rating'] = pd.to_numeric(df['rating'].astype(str).str.replace('|', '', regex=False), errors='coerce')


df = df.dropna(subset=['rating'])

df['discounted_price'] = df['discounted_price'].astype(str).str.replace('₹', '').str.replace(',', '').astype(float)
df['actual_price'] = df['actual_price'].astype(str).str.replace('₹', '').str.replace(',', '').astype(float)
df['discount_percentage'] = df['discount_percentage'].astype(str).str.replace('%', '').astype(float) / 100.0


df['rating_count'] = df['rating_count'].astype(str).str.replace(',', '')
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')
df['rating_count'] = df['rating_count'].fillna(df['rating_count'].median())


df['review_content'] = df['review_content'].fillna("")
df['review_title'] = df['review_title'].fillna("")
df['product_name'] = df['product_name'].fillna("")

df['full_review'] = df['review_title'] + " " + df['review_content']

print("\nPost-processing Info:")
print(df.info())
print(df[['rating', 'discounted_price', 'actual_price', 'rating_count']].describe())
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.layers import Input, Embedding, LSTM, Dense, Dropout, concatenate
from tensorflow.keras.models import Model
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

max_vocab = 10000
max_len = 150

tokenizer = Tokenizer(num_words=max_vocab, oov_token="<OOV>")
tokenizer.fit_on_texts(df['full_review'])

text_sequences = tokenizer.texts_to_sequences(df['full_review'])
padded_sequences = pad_sequences(text_sequences, maxlen=max_len, padding='post', truncating='post')


scaler = StandardScaler()
num_features = df[['discounted_price', 'actual_price', 'discount_percentage', 'rating_count']].values
scaled_num_features = scaler.fit_transform(num_features)


X_text_train, X_text_val, X_num_train, X_num_val, y_train, y_val = train_test_split(
    padded_sequences, scaled_num_features, df['rating'].values, test_size=0.2, random_state=42
)


text_input = Input(shape=(max_len,), name='text_input')
embedding = Embedding(input_dim=max_vocab, output_dim=64, input_length=max_len)(text_input)
lstm_out = LSTM(32, dropout=0.2, recurrent_dropout=0.2)(embedding)

num_input = Input(shape=(scaled_num_features.shape[1],), name='num_input')
num_dense = Dense(16, activation='relu')(num_input)


merged = concatenate([lstm_out, num_dense])
fc1 = Dense(16, activation='relu')(merged)
dropout = Dropout(0.2)(fc1)
output = Dense(1, activation='linear', name='rating_output')(dropout)

model = Model(inputs=[text_input, num_input], outputs=output)
model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.summary()


history = model.fit(
    [X_text_train, X_num_train], y_train,
    validation_data=([X_text_val, X_num_val], y_val),
    epochs=10,
    batch_size=32,
    verbose=1
)


loss, mae = model.evaluate([X_text_val, X_num_val], y_val, verbose=0)
print(f"\nEvaluation results - Validation Loss (MSE): {loss:.4f}, Mean Absolute Error (MAE): {mae:.4f}")
```

```python

import numpy as np


val_predictions = model.predict([X_text_val, X_num_val]).flatten()
comparison_df = pd.DataFrame({
    'Actual Rating': y_val,
    'Predicted Rating': val_predictions
})
print("\nSample Comparisons (Actual vs Predicted):")
print(comparison_df.head(10))

plt.figure(figsize=(10, 4))
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Model training history')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.show()

unseen_reviews = [
    "This product is absolutely amazing! Worked flawlessly and looks highly durable.",
    "Terrible quality, broke after first use. Do not recommend."
]
unseen_seqs = tokenizer.texts_to_sequences(unseen_reviews)
unseen_padded = pad_sequences(unseen_seqs, maxlen=max_len, padding='post', truncating='post')


dummy_num_features = np.zeros((len(unseen_reviews), num_features.shape[1]))

predicted_unseen = model.predict([unseen_padded, dummy_num_features]).flatten()
for review, pred in zip(unseen_reviews, predicted_unseen):
    print(f"\nReview: \"{review}\"")
    print(f"Predicted Customer Rating: {pred:.2f}/5.0")
```
